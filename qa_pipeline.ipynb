```python
import pandas as pd
from src.pipeline import RAGPipeline
from src.vector_store import VectorStore
from src.generator import AnswerGenerator
from src.utils import CorpusLoader
from src.chunkers import FixedChunker, NltkSemanticChunker, RecursiveChunker

import os 
import json
from dotenv import load_dotenv
load_dotenv()
```

```python
df = pd.read_json('data/MultiHopRAG.json')
df
```

```python
df['fact_lens'] = df['evidence_list'].apply(lambda x: [len(i['fact']) for i in x])
df['fact_lens'].apply(lambda x: max(x) if len(x) > 0 else None).max()
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')

fixed_pipeline = RAGPipeline(api_key=groq_api_key, 
                            chunker=FixedChunker(chunk_size=512),
                            vector_store=VectorStore(persist_directory='./data/fixed_chunkers_db'),
                            document_loader=CorpusLoader())
fixed_pipeline.build_from_document(search_kwargs={'k': 4})
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')

recursive_pipeline = RAGPipeline(api_key=groq_api_key, 
                            chunker=RecursiveChunker(chunk_size=512),
                            vector_store=VectorStore(persist_directory='./data/recursive_chunkers_db'),
                            document_loader=CorpusLoader())
recursive_pipeline.build_from_document(search_kwargs={'k': 4})
```

```python
groq_api_key = os.getenv('GROQ_API_KEY')

semantic_pipeline = RAGPipeline(api_key=groq_api_key, 
                            chunker=NltkSemanticChunker(chunk_size=512),
                            vector_store=VectorStore(persist_directory='./data/semantic_chunkers_db'),
                            document_loader=CorpusLoader())
semantic_pipeline.build_from_document(search_kwargs={'k': 4})
```

```python
questions = list(df['query'][0:100])
lst = []
for i, question in enumerate(questions):
    print(f"Вопрос: {i}")
    answer = fixed_pipeline.query(question)
    lst.append(answer)
```

```python
questions = list(df['query'][0:100])
semantic_answers = []
for i, question in enumerate(questions):
    print(f"Вопрос: {i}")
    answer = semantic_pipeline.query(question)
    semantic_answers.append(answer)
```

```python
questions = list(df['query'][0:100])
recursive_answers = []
for i, question in enumerate(questions):
    print(f"Вопрос: {i}")
    answer = recursive_pipeline.query(question)
    recursive_answers.append(answer)
```

```python
fixed_results_qa = pd.DataFrame(lst)
semantic_results_qa = pd.DataFrame(semantic_answers)
recursive_results_qa = pd.DataFrame(recursive_answers)

fixed_results_qa.to_csv('results/fixed_results_qa_part_1.csv', index=False)
semantic_results_qa.to_csv('results/semantic_results_qa_part_1.csv', index=False)
recursive_results_qa.to_csv('results/recursive_results_qa_part_1.csv', index=False)
```

```python
questions_2 = list(df['query'][100:200])

fixed_answers = []
recursive_answers = []
semantic_answers = []
for i, question in enumerate(questions_2):
    print(f"Вопрос: {i}")
    fixed_answer = fixed_pipeline.query(question)
    fixed_answers.append(fixed_answer)

    recursive_answer = recursive_pipeline.query(question)
    recursive_answers.append(recursive_answer)

    semantic_answer = semantic_pipeline.query(question)
    semantic_answers.append(semantic_answer)

fixed_results_qa = pd.DataFrame(fixed_answers)
semantic_results_qa = pd.DataFrame(semantic_answers)
recursive_results_qa = pd.DataFrame(recursive_answers)

fixed_results_qa.to_csv('results/fixed_results_qa_part_2.csv', index=False)
semantic_results_qa.to_csv('results/semantic_results_qa_part_2.csv', index=False)
recursive_results_qa.to_csv('results/recursive_results_qa_part_2.csv', index=False)
```
